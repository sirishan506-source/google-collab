**Import required libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

**Load the IRIS dataset**

```python
import pandas as pd

df = pd.read_csv('/content/Iris.csv')

print(df.head())
print(df.shape)
```

**Separate features and target**

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

print(X.head())
print(y.head())
```

**Standardize the data**

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

**Apply PCA with 2 Principal Components**

```python
pca2 = PCA(n_components=2)

X_pca2 = pca2.fit_transform(X_scaled)

print(X_pca2[:5])
```

**Find explained variance for 2 PCs**

```python
print("Explained variance ratio:")
print(pca2.explained_variance_ratio_)

print("Total explained variance:")
print(sum(pca2.explained_variance_ratio_))
```

**Plot 2 Principal Components**

```python
plt.figure(figsize=(8,5))

for species in y.unique():
    plt.scatter(
        X_pca2[y == species, 0],
        X_pca2[y == species, 1],
        label=species
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("IRIS Dataset - PCA with 2 Components")
plt.legend()
plt.show()
```

**Apply PCA with 3 Principal Components**

```python
pca3 = PCA(n_components=3)

X_pca3 = pca3.fit_transform(X_scaled)

print(X_pca3[:5])
```

**Find explained variance for 3 PCs**

```python
print("Explained variance ratio:")
print(pca3.explained_variance_ratio_)

print("Total explained variance:")
print(sum(pca3.explained_variance_ratio_))
```

**Plot 3 Principal Components**

```python
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

for species in y.unique():
    ax.scatter(
        X_pca3[y == species, 0],
        X_pca3[y == species, 1],
        X_pca3[y == species, 2],
        label=species
    )

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

ax.set_title("IRIS Dataset - PCA with 3 Components")
ax.legend()

plt.show()
```

**PCA component loadings**

```python
components = pd.DataFrame(
    pca3.components_,
    columns=X.columns,
    index=['PC1', 'PC2', 'PC3']
)

print(components)
```
