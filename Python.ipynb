Changing Item value

Inserting items with the use of insert method

to add an item to the end of the list we use append()

To append elements from one list to another we use extend()

For removing the element from the list we use remove()

for removing specified item we use pop() with number

delete is a keyword for deleting whole list

to clear list we use clear() which makes empty list

```python
lst = ['apple','Ball','Cat']
lst[1]='Bat'
print(lst)
lst.insert(2,'Dog')
print(lst)
lst.append('Elephant')
print(lst)
lst2 = ['Fan','goal','Heaven']
lst.extend(lst2)
print(lst)
lst.remove("Heaven")
print(lst)
lst.pop(1)
print(lst)
```

# List

Printing all items in list one by one

```python
lst = ['apple','Ball','Cat']
for x in lst:
  print(x)
```

loop through the index number

```python
lst=["Apple","banana","cat"]
for x in range(len(lst)):
  print(lst[x])
```

While loop

```python
lst=["Apple","banana","cat"]
i = 0
while(i<len(lst)):
  print(lst[i])
  i = i+1
```

Looping using list comprehension

```python
lst=["Apple","banana","cat"]
[print(x) for x in lst]
```

**Syntax for list comprehension** :
newlist = [expression for item in iterable if condition == True]

expression → What to add to the list (x)

item → Variable (x)

iterable → Sequence (range(10))

condition → Filter (x < 5)

```python
newlist=[x for x in range(10) if x<=5]
print(newlist)
```

```python
lst=["Apple","banana","cat"]
newlist=[x.upper() for x in lst]
print(newlist)
```

Sorting list items

```python
lst = ['cat','apple','ball','elephant']
lst.sort()
print(lst)
```

sort descending

```python
lst = [20,40,10,3,5,2]
lst.sort(reverse = True)
print(lst)
```

```python
lst = [20,40,10,3,5,2]
lst.sort(reverse = False)
print(lst)
```

clear()	    Removes all the elements from the list
